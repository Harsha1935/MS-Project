```python
import os
import shutil
import time

from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm.auto import tqdm


DRIVE_IMAGE_ROOT = (
    "/content/drive/MyDrive/"
    "Dataset_extracted/colored_images/colored_images"
)

LOCAL_IMAGE_ROOT = "/content/dr_images"

COPY_COMPLETE_FILE = "/content/.dr_images_copy_complete"

MAX_WORKERS = 8
MAX_RETRIES = 5


os.makedirs(
    LOCAL_IMAGE_ROOT,
    exist_ok=True
)


if not os.path.exists(DRIVE_IMAGE_ROOT):
    raise RuntimeError(
        f"Google Drive folder not found:\n{DRIVE_IMAGE_ROOT}"
    )


print("Scanning files...")


all_files = []

for root, dirs, filenames in os.walk(
    DRIVE_IMAGE_ROOT
):

    for filename in filenames:

        src = os.path.join(
            root,
            filename
        )

        relative_path = os.path.relpath(
            src,
            DRIVE_IMAGE_ROOT
        )

        dst = os.path.join(
            LOCAL_IMAGE_ROOT,
            relative_path
        )

        all_files.append(
            (src, dst)
        )


print(
    f"Total files: {len(all_files):,}"
)


remaining_files = []


for src, dst in tqdm(
    all_files,
    desc="Checking existing files",
    unit="file"
):

    try:

        if os.path.exists(dst):

            src_size = os.path.getsize(src)
            dst_size = os.path.getsize(dst)

            if src_size == dst_size:
                continue

            os.remove(dst)

        remaining_files.append(
            (src, dst)
        )

    except Exception:

        remaining_files.append(
            (src, dst)
        )


already_complete = (
    len(all_files)
    -
    len(remaining_files)
)


print(
    f"\nAlready copied correctly: {already_complete:,}"
)

print(
    f"Remaining to copy: {len(remaining_files):,}"
)


def copy_with_retry(item):

    src, dst = item

    os.makedirs(
        os.path.dirname(dst),
        exist_ok=True
    )

    last_error = None


    for attempt in range(
        1,
        MAX_RETRIES + 1
    ):

        try:

            if os.path.exists(dst):

                try:

                    if (
                        os.path.getsize(src)
                        ==
                        os.path.getsize(dst)
                    ):
                        return True, src, None

                except Exception:
                    pass


                try:
                    os.remove(dst)
                except Exception:
                    pass


            shutil.copyfile(
                src,
                dst
            )


            if (
                os.path.getsize(src)
                !=
                os.path.getsize(dst)
            ):

                raise IOError(
                    "Copied file size does not match source."
                )


            return True, src, None


        except Exception as e:

            last_error = e


            try:
                if os.path.exists(dst):
                    os.remove(dst)
            except Exception:
                pass


            time.sleep(
                min(
                    attempt * 2,
                    10
                )
            )


    return (
        False,
        src,
        str(last_error)
    )


errors = []


with ThreadPoolExecutor(
    max_workers=MAX_WORKERS
) as executor:

    futures = [
        executor.submit(
            copy_with_retry,
            item
        )
        for item in remaining_files
    ]


    for future in tqdm(
        as_completed(futures),
        total=len(futures),
        desc="Copying DR images",
        unit="file"
    ):

        success, src, error = (
            future.result()
        )


        if not success:

            errors.append(
                {
                    "file": src,
                    "error": error
                }
            )


print("\n" + "=" * 70)
print("COPY CHECK")
print("=" * 70)


if errors:

    print(
        f"Files still failed after retries: "
        f"{len(errors):,}"
    )

    print("\nFirst few failed files:")

    for item in errors[:10]:

        print(
            item["file"]
        )

        print(
            item["error"]
        )

    print(
        "\nRun this same cell again. "
        "It will skip all successfully copied files "
        "and retry only the remaining files."
    )

else:

    with open(
        COPY_COMPLETE_FILE,
        "w"
    ) as f:

        f.write("complete")


    print(
        "All DR images copied successfully."
    )

    print(
        "Local folder:",
        LOCAL_IMAGE_ROOT
    )
```

```python
# DATASET + DATALOADERS

import random as _random
import torchvision.transforms.functional as TF


_MEAN = np.array(
    [0.485, 0.456, 0.406],
    dtype=np.float32
)

_STD = np.array(
    [0.229, 0.224, 0.225],
    dtype=np.float32
)


class DRStage3Dataset(Dataset):

    def __init__(
        self,
        df,
        size=384,
        augment=False
    ):
        self.df = df.reset_index(drop=True)
        self.size = size
        self.augment = augment


    def __len__(self):
        return len(self.df)


    def _load_rgb(self, path):

        img = Image.open(path).convert("RGB")

        img = img.resize(
            (self.size, self.size),
            Image.BILINEAR
        )

        return img


    def _load_mask(self, path):

        mask = Image.open(path).convert("L")

        mask = mask.resize(
            (self.size, self.size),
            Image.NEAREST
        )

        return mask


    def _augment(
        self,
        img_pil,
        mask_pil
    ):

        # Horizontal flip
        if _random.random() > 0.5:

            img_pil = TF.hflip(img_pil)

            mask_pil = TF.hflip(mask_pil)


        # Vertical flip
        if _random.random() > 0.5:

            img_pil = TF.vflip(img_pil)

            mask_pil = TF.vflip(mask_pil)


        # Rotation
        k = _random.randint(0, 3)

        if k > 0:

            img_pil = TF.rotate(
                img_pil,
                k * 90
            )

            mask_pil = TF.rotate(
                mask_pil,
                k * 90
            )


        # Brightness
        if _random.random() > 0.4:

            img_pil = TF.adjust_brightness(
                img_pil,
                1.0 + _random.uniform(-0.3, 0.3)
            )


        # Contrast
        if _random.random() > 0.4:

            img_pil = TF.adjust_contrast(
                img_pil,
                1.0 + _random.uniform(-0.3, 0.3)
            )


        # Saturation
        if _random.random() > 0.4:

            img_pil = TF.adjust_saturation(
                img_pil,
                1.0 + _random.uniform(-0.3, 0.3)
            )


        return img_pil, mask_pil


    def _to_tensor(
        self,
        img_pil,
        mask_pil
    ):

        # RGB image
        arr = (
            np.array(img_pil)
            .astype(np.float32)
            / 255.0
        )

        arr = (
            arr - _MEAN
        ) / _STD


        rgb = torch.tensor(
            np.transpose(
                arr,
                (2, 0, 1)
            ),
            dtype=torch.float32
        )


        # Stage 3 mask
        m = (
            np.array(mask_pil)
            .astype(np.float32)
            / 255.0
            > 0.5
        ).astype(np.float32)


        mask = torch.tensor(
            m[None, :, :],
            dtype=torch.float32
        )


        return rgb, mask


    def __getitem__(self, idx):

        row = self.df.iloc[idx]


        # Original RGB image
        img_pil = self._load_rgb(
            str(row["img_path"])
        )


        # STRICTLY STAGE 3 MASK
        mask_pil = self._load_mask(
            str(row["stage3_mask_path"])
        )


        label = int(
            row["label"]
        )


        if self.augment:

            img_pil, mask_pil = self._augment(
                img_pil,
                mask_pil
            )


        rgb, mask = self._to_tensor(
            img_pil,
            mask_pil
        )


        return rgb, mask, label


# --------------------------------------------------
# OVERSAMPLING
# --------------------------------------------------

# Mild 2x, severe 2x, PDR 2x
mild_df = train_df[
    train_df["label"] == 1
]

severe_df = train_df[
    train_df["label"] == 3
]

pdr_df = train_df[
    train_df["label"] == 4
]


train_df_balanced = pd.concat(
    [
        train_df,
        mild_df,
        severe_df,
        pdr_df,
    ],
    ignore_index=True
)


print(
    "Balanced train class counts:"
)

vc = (
    train_df_balanced["label"]
    .value_counts()
    .sort_index()
)


for i, cn in enumerate(CLASS_NAMES):

    n = vc.get(i, 0)

    print(
        f"  {cn:<22}: {n:>5}"
    )


print(
    f"\nTrain original : "
    f"{len(train_df)}"
)

print(
    f"Train balanced : "
    f"{len(train_df_balanced)}"
)

print(
    f"Val            : "
    f"{len(val_df)}"
)

print(
    f"Test           : "
    f"{len(test_df)}"
)


# --------------------------------------------------
# DATASETS
# --------------------------------------------------

stage3_train_ds = DRStage3Dataset(
    train_df_balanced,
    size=STAGE3_SIZE,
    augment=True
)

stage3_val_ds = DRStage3Dataset(
    val_df,
    size=STAGE3_SIZE,
    augment=False
)

stage3_test_ds = DRStage3Dataset(
    test_df,
    size=STAGE3_SIZE,
    augment=False
)


# --------------------------------------------------
# DATALOADERS
# --------------------------------------------------

dr_train_loader = DataLoader(
    stage3_train_ds,
    batch_size=STAGE3_BATCH,
    shuffle=True,
    num_workers=STAGE3_NUM_WORKERS,
    pin_memory=PIN_MEMORY,
    persistent_workers=True,
    prefetch_factor=2
)


dr_val_loader = DataLoader(
    stage3_val_ds,
    batch_size=STAGE3_BATCH,
    shuffle=False,
    num_workers=STAGE3_NUM_WORKERS,
    pin_memory=PIN_MEMORY,
    persistent_workers=True,
    prefetch_factor=2
)


dr_test_loader = DataLoader(
    stage3_test_ds,
    batch_size=STAGE3_BATCH,
    shuffle=False,
    num_workers=STAGE3_NUM_WORKERS,
    pin_memory=PIN_MEMORY,
    persistent_workers=True,
    prefetch_factor=2
)


print(
    f"\nTrain batches  : "
    f"{len(dr_train_loader)}"
)

print(
    f"Val batches    : "
    f"{len(dr_val_loader)}"
)

print(
    f"Test batches   : "
    f"{len(dr_test_loader)}"
)
```

```python
import numpy as np

import torch
import torch.nn as nn

import timm

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    recall_score,
    precision_score,
    roc_auc_score
)

from sklearn.preprocessing import label_binarize


RGB_SCALE = 0.75
MASK_SCALE = 1.25
DROP = 0.45


if "AMP_DEVICE" not in globals():

    AMP_DEVICE = (
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )


def macro_auc(
    y_true,
    probabilities
):

    try:

        y_binary = label_binarize(
            y_true,
            classes=list(
                range(
                    num_classes
                )
            )
        )

        return roc_auc_score(
            y_binary,
            probabilities,
            average="macro",
            multi_class="ovr"
        )

    except Exception:

        return np.nan


class SEBlock(nn.Module):

    def __init__(
        self,
        in_dim,
        reduction=16
    ):

        super().__init__()

        hidden_dim = max(
            1,
            in_dim // reduction
        )

        self.fc = nn.Sequential(

            nn.Linear(
                in_dim,
                hidden_dim,
                bias=False
            ),

            nn.ReLU(
                inplace=True
            ),

            nn.Linear(
                hidden_dim,
                in_dim,
                bias=False
            ),

            nn.Sigmoid()
        )


    def forward(
        self,
        x
    ):

        weights = self.fc(
            x
        )

        return (
            x
            *
            weights
        )


class _MaskEncoderProxy:

    def __init__(
        self,
        model
    ):

        self.enc3 = (
            model.blocks[-2]
        )

        self.enc4 = (
            model.blocks[-1]
        )


class Stage3EfficientNetB0BxSE(nn.Module):

    def __init__(
        self,
        num_classes,
        stage1_weight_path,
        mask_model_name,
        mask_pretrained=True
    ):

        super().__init__()


        self.rgb_scale = (
            RGB_SCALE
        )

        self.mask_scale = (
            MASK_SCALE
        )

        self.mask_model_name = (
            mask_model_name
        )


        rgb_model = timm.create_model(
            "tf_efficientnet_b0.ns_jft_in1k",
            pretrained=False,
            num_classes=0,
            global_pool="avg"
        )


        stage1_state = torch.load(
            stage1_weight_path,
            map_location="cpu"
        )


        if (
            isinstance(
                stage1_state,
                dict
            )
            and
            "state_dict"
            in stage1_state
        ):

            stage1_state = (
                stage1_state[
                    "state_dict"
                ]
            )


        if (
            isinstance(
                stage1_state,
                dict
            )
            and
            "model_state_dict"
            in stage1_state
        ):

            stage1_state = (
                stage1_state[
                    "model_state_dict"
                ]
            )


        stage1_state = {

            key:
                value

            for (
                key,
                value
            )
            in stage1_state.items()

            if not key.startswith(
                "classifier"
            )
        }


        rgb_model.load_state_dict(
            stage1_state,
            strict=False
        )


        for parameter in (
            rgb_model.parameters()
        ):

            parameter.requires_grad = False


        self.rgb_branch = (
            rgb_model
        )


        mask_model = timm.create_model(
            mask_model_name,
            pretrained=mask_pretrained,
            num_classes=0,
            global_pool="avg"
        )


        old_conv = (
            mask_model.conv_stem
        )


        new_conv = nn.Conv2d(
            1,
            old_conv.out_channels,
            kernel_size=
                old_conv.kernel_size,
            stride=
                old_conv.stride,
            padding=
                old_conv.padding,
            dilation=
                old_conv.dilation,
            groups=
                old_conv.groups,
            bias=False
        )


        with torch.no_grad():

            new_conv.weight.copy_(
                old_conv.weight.mean(
                    dim=1,
                    keepdim=True
                )
            )


        mask_model.conv_stem = (
            new_conv
        )


        if hasattr(
            mask_model,
            "set_grad_checkpointing"
        ):

            try:

                mask_model.set_grad_checkpointing(
                    True
                )

            except Exception:

                pass


        self.mask_branch = (
            mask_model
        )


        self._mask_encoder_proxy = (
            _MaskEncoderProxy(
                mask_model
            )
        )


        self.RGB_DIM = int(
            rgb_model.num_features
        )


        self.MASK_DIM = int(
            mask_model.num_features
        )


        self.FUSE_DIM = (
            self.RGB_DIM
            +
            self.MASK_DIM
        )


        self.se = SEBlock(
            self.FUSE_DIM,
            reduction=16
        )


        self.head = nn.Sequential(

            nn.Dropout(
                DROP
            ),

            nn.Linear(
                self.FUSE_DIM,
                num_classes
            )
        )


    @property
    def mask_encoder(
        self
    ):

        return (
            self._mask_encoder_proxy
        )


    def _features(
        self,
        rgb,
        mask
    ):

        with torch.no_grad():

            rgb_features = (
                self.rgb_branch(
                    rgb
                )
            )

        mask_features = (
            self.mask_branch(
                mask
            )
        )


        rgb_features = (
            self.rgb_scale
            *
            rgb_features
        )


        mask_features = (
            self.mask_scale
            *
            mask_features
        )


        fused_features = torch.cat(
            [
                rgb_features,
                mask_features
            ],
            dim=1
        )


        attended_features = (
            self.se(
                fused_features
            )
        )


        return (
            attended_features,
            rgb_features,
            mask_features
        )


    def forward(
        self,
        rgb,
        mask
    ):

        (
            attended_features,
            _,
            _
        ) = self._features(
            rgb,
            mask
        )


        logits = self.head(
            attended_features
        )


        return logits


    def forward_with_parts(
        self,
        rgb,
        mask
    ):

        (
            attended_features,
            rgb_features,
            mask_features
        ) = self._features(
            rgb,
            mask
        )


        logits = self.head(
            attended_features
        )


        return (
            logits,
            mask_features
        )


    def forward_rgb_only(
        self,
        rgb,
        mask=None
    ):

        with torch.no_grad():

            rgb_features = (
                self.rgb_scale
                *
                self.rgb_branch(
                    rgb
                )
            )


        zero_mask_features = (
            torch.zeros(
                rgb_features.size(0),
                self.MASK_DIM,
                device=
                    rgb_features.device,
                dtype=
                    rgb_features.dtype
            )
        )


        fused = torch.cat(
            [
                rgb_features,
                zero_mask_features
            ],
            dim=1
        )


        attended = self.se(
            fused
        )


        return self.head(
            attended
        )


    def forward_mask_only(
        self,
        rgb,
        mask
    ):

        mask_features = (
            self.mask_scale
            *
            self.mask_branch(
                mask
            )
        )


        zero_rgb_features = (
            torch.zeros(
                mask_features.size(0),
                self.RGB_DIM,
                device=
                    mask_features.device,
                dtype=
                    mask_features.dtype
            )
        )


        fused = torch.cat(
            [
                zero_rgb_features,
                mask_features
            ],
            dim=1
        )


        attended = self.se(
            fused
        )


        return self.head(
            attended
        )


@torch.no_grad()
def evaluate(
    model,
    loader
):

    model.eval()


    labels_all = []

    predictions_all = []

    probabilities_all = []

    validation_losses = []


    for (
        rgb,
        mask,
        labels
    ) in loader:


        rgb = rgb.to(
            device,
            non_blocking=True
        )


        mask = mask.to(
            device,
            non_blocking=True
        )


        labels = labels.to(
            device,
            non_blocking=True
        )


        if torch.cuda.is_available():

            rgb = rgb.contiguous(
                memory_format=
                    torch.channels_last
            )


            mask = mask.contiguous(
                memory_format=
                    torch.channels_last
            )


        with torch.amp.autocast(
            device_type=
                AMP_DEVICE,
            enabled=
                torch.cuda.is_available()
        ):

            logits = model(
                rgb,
                mask
            )


            loss = criterion(
                logits,
                labels
            )


            probabilities = (
                torch.softmax(
                    logits,
                    dim=1
                )
            )


        validation_losses.append(
            float(
                loss
                .detach()
                .cpu()
                .item()
            )
        )


        labels_all.extend(
            labels
            .detach()
            .cpu()
            .numpy()
        )


        predictions_all.extend(
            logits
            .argmax(
                dim=1
            )
            .detach()
            .cpu()
            .numpy()
        )


        probabilities_all.extend(
            probabilities
            .detach()
            .cpu()
            .numpy()
        )


    labels_all = np.asarray(
        labels_all
    )


    predictions_all = np.asarray(
        predictions_all
    )


    probabilities_all = np.asarray(
        probabilities_all
    )


    accuracy = accuracy_score(
        labels_all,
        predictions_all
    )


    macro_f1 = f1_score(
        labels_all,
        predictions_all,
        average="macro",
        zero_division=0
    )


    weighted_f1 = f1_score(
        labels_all,
        predictions_all,
        average="weighted",
        zero_division=0
    )


    auc_value = macro_auc(
        labels_all,
        probabilities_all
    )


    recalls = recall_score(
        labels_all,
        predictions_all,
        average=None,
        labels=list(
            range(
                num_classes
            )
        ),
        zero_division=0
    )


    precisions = precision_score(
        labels_all,
        predictions_all,
        average=None,
        labels=list(
            range(
                num_classes
            )
        ),
        zero_division=0
    )


    return {

        "labels":
            labels_all,

        "preds":
            predictions_all,

        "probs":
            probabilities_all,

        "loss":
            float(
                np.mean(
                    validation_losses
                )
            ),

        "accuracy":
            accuracy,

        "macro_f1":
            macro_f1,

        "weighted_f1":
            weighted_f1,

        "macro_auc":
            auc_value,

        "recall":
            recalls,

        "precision":
            precisions
    }


print(
    "Stage3EfficientNetB0BxSE defined:",
    "Stage3EfficientNetB0BxSE"
    in globals()
)


print(
    "evaluate defined:",
    "evaluate"
    in globals()
)
```

```python
import os
import gc
import math
import random
import shutil
import warnings

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.optim as optim
import timm

from torch.utils.data import DataLoader

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    recall_score,
    precision_score,
    roc_auc_score
)

from sklearn.preprocessing import label_binarize
from tqdm.auto import tqdm

warnings.filterwarnings(
    "ignore",
    message="name."
)

try:
    from huggingface_hub import logging as hf_logging
    hf_logging.set_verbosity_error()
except Exception:
    pass

EPOCHS = 60
PATIENCE = 5
EFFECTIVE_BATCH = 64

LR = 1e-4
WEIGHT_DECAY = 1e-3

AUX_WEIGHT = 0.40
MASK_DROP_P = 0.10

SEED_COMPARE = 42

# Fixed B3 backbone for the GAN-refinement ablation; original timm architecture.
MODELS = {
    "B3": "tf_efficientnet_b3.ns_jft_in1k"
}

BATCH_BY_MODEL = {
    "B3": 16
}

if "FAST_NUM_WORKERS" not in globals():
    cpu_count = os.cpu_count() or 2
    FAST_NUM_WORKERS = min(
        4,
        max(
            2,
            cpu_count - 1
        )
    )

if "FAST_PIN_MEMORY" not in globals():
    FAST_PIN_MEMORY = torch.cuda.is_available()

if "USE_AMP" not in globals():
    USE_AMP = torch.cuda.is_available()

if "AMP_DEVICE" not in globals():
    AMP_DEVICE = (
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )

required_items = [
    "stage3_train_ds",
    "stage3_val_ds",
    "Stage3EfficientNetB0BxSE",
    "device",
    "num_classes",
    "CLASS_NAMES",
    "STAGE1_W"
]

missing = [
    name
    for name in required_items
    if name not in globals()
]

if missing:
    raise RuntimeError(
        "Missing required variables: "
        + ", ".join(missing)
        + "\nRun your dataset/model setup cells first."
    )

def seed_all_compare():
    random.seed(SEED_COMPARE)
    np.random.seed(SEED_COMPARE)
    torch.manual_seed(SEED_COMPARE)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED_COMPARE)

DRIVE_COMPARE_DIR = (
    "/content/drive/MyDrive/DR_Project/"
    "stage3_ablation_B3_refined_mask_SE"
)

LOCAL_COMPARE_DIR = (
    "/content/stage3_ablation_B3_refined_mask_SE"
)

os.makedirs(
    DRIVE_COMPARE_DIR,
    exist_ok=True
)

os.makedirs(
    LOCAL_COMPARE_DIR,
    exist_ok=True
)

print("Drive comparison directory:")
print(DRIVE_COMPARE_DIR)

LOCAL_STAGE1_W = os.path.join(
    "/content",
    os.path.basename(STAGE1_W)
)

if not os.path.exists(LOCAL_STAGE1_W):
    shutil.copy2(
        STAGE1_W,
        LOCAL_STAGE1_W
    )

class_weights = torch.tensor(
    [
        0.65,
        2.20,
        1.15,
        2.20,
        1.80
    ],
    dtype=torch.float32,
    device=device
)

criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=0.05
)

aux_criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=0.05
)

def macro_auc_with_loss(
    y_true,
    probabilities
):
    try:
        y_binary = label_binarize(
            y_true,
            classes=list(
                range(num_classes)
            )
        )

        return float(
            roc_auc_score(
                y_binary,
                probabilities,
                average="macro",
                multi_class="ovr"
            )
        )

    except Exception:
        return float("nan")

@torch.no_grad()
def evaluate_with_loss(
    model,
    loader
):
    model.eval()

    labels_all = []
    predictions_all = []
    probabilities_all = []
    validation_losses = []

    for rgb, mask, labels in loader:
        rgb = rgb.to(
            device,
            non_blocking=True
        )

        mask = mask.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        if torch.cuda.is_available():
            rgb = rgb.contiguous(
                memory_format=torch.channels_last
            )

            mask = mask.contiguous(
                memory_format=torch.channels_last
            )

        with torch.amp.autocast(
            device_type=AMP_DEVICE,
            enabled=USE_AMP
        ):
            logits = model(
                rgb,
                mask
            )

            loss = criterion(
                logits,
                labels
            )

        probabilities = torch.softmax(
            logits,
            dim=1
        )

        predictions = logits.argmax(
            dim=1
        )

        validation_losses.append(
            float(
                loss.detach()
                .cpu()
                .item()
            )
        )

        labels_all.extend(
            labels.detach()
            .cpu()
            .numpy()
        )

        predictions_all.extend(
            predictions.detach()
            .cpu()
            .numpy()
        )

        probabilities_all.extend(
            probabilities.detach()
            .cpu()
            .numpy()
        )

    labels_all = np.asarray(
        labels_all
    )

    predictions_all = np.asarray(
        predictions_all
    )

    probabilities_all = np.asarray(
        probabilities_all
    )

    accuracy = accuracy_score(
        labels_all,
        predictions_all
    )

    macro_f1_value = f1_score(
        labels_all,
        predictions_all,
        average="macro",
        zero_division=0
    )

    weighted_f1_value = f1_score(
        labels_all,
        predictions_all,
        average="weighted",
        zero_division=0
    )

    macro_auc_value = macro_auc_with_loss(
        labels_all,
        probabilities_all
    )

    recalls = recall_score(
        labels_all,
        predictions_all,
        average=None,
        labels=list(
            range(num_classes)
        ),
        zero_division=0
    )

    precisions = precision_score(
        labels_all,
        predictions_all,
        average=None,
        labels=list(
            range(num_classes)
        ),
        zero_division=0
    )

    return {
        "labels": labels_all,
        "preds": predictions_all,
        "probs": probabilities_all,
        "loss": float(
            np.mean(
                validation_losses
            )
        ),
        "accuracy": float(accuracy),
        "macro_f1": float(macro_f1_value),
        "weighted_f1": float(weighted_f1_value),
        "macro_auc": float(macro_auc_value),
        "recall": recalls,
        "precision": precisions
    }

def make_train_val_loaders(
    batch_size
):
    common = {
        "num_workers": FAST_NUM_WORKERS,
        "pin_memory": FAST_PIN_MEMORY
    }

    if FAST_NUM_WORKERS > 0:
        common[
            "persistent_workers"
        ] = True

        common[
            "prefetch_factor"
        ] = 4

    generator = torch.Generator()

    generator.manual_seed(
        SEED_COMPARE
    )

    train_loader = DataLoader(
        stage3_train_ds,
        batch_size=batch_size,
        shuffle=True,
        generator=generator,
        **common
    )

    val_loader = DataLoader(
        stage3_val_ds,
        batch_size=batch_size,
        shuffle=False,
        **common
    )

    return (
        train_loader,
        val_loader
    )

def print_saved_training_history(
    short_name
):
    history_path = os.path.join(
        DRIVE_COMPARE_DIR,
        f"stage3_{short_name}_history.csv"
    )

    if not os.path.exists(
        history_path
    ):
        print(
            f"History not found for {short_name}."
        )

        return

    history_df = pd.read_csv(
        history_path
    )

    if len(history_df) == 0:
        print(
            f"History empty for {short_name}."
        )

        return

    print(
        "\n"
        + "=" * 90
    )

    print(
        f"SAVED TRAINING HISTORY: {short_name}"
    )

    print(
        "=" * 90
    )

    best_f1 = -1.0
    wait = 0

    for _, row in history_df.iterrows():
        epoch = int(
            row[
                "epoch"
            ]
        )

        val_f1 = float(
            row[
                "val_macro_f1"
            ]
        )

        print(
            f"{short_name} | "
            f"Epoch {epoch:02d}/{EPOCHS} | "
            f"Train Total Loss={float(row['train_loss']):.4f} | "
            f"Train Main Loss={float(row['train_main_loss']):.4f} | "
            f"Val Loss={float(row['val_loss']):.4f} | "
            f"Val Accuracy={float(row['val_accuracy']):.4f} | "
            f"Val Macro-F1={val_f1:.4f} | "
            f"Val AUROC={float(row['val_macro_auc']):.4f}"
        )

        if val_f1 > best_f1:
            best_f1 = val_f1
            wait = 0

            print(
                "Best validation Macro-F1 checkpoint saved."
            )

        else:
            wait += 1

            print(
                f"Early stopping patience: "
                f"{wait}/{PATIENCE}"
            )

            if wait >= PATIENCE:
                print(
                    f"{short_name}: Early stopping."
                )

                break

def history_completed(
    short_name
):
    history_path = os.path.join(
        DRIVE_COMPARE_DIR,
        f"stage3_{short_name}_history.csv"
    )

    if not os.path.exists(
        history_path
    ):
        return False

    try:
        history_df = pd.read_csv(
            history_path
        )

        if len(history_df) == 0:
            return False

        required_history_columns = {
            "train_main_loss",
            "val_loss"
        }

        if not required_history_columns.issubset(
            history_df.columns
        ):
            return False

        best_f1 = -1.0
        wait = 0

        for _, row in history_df.iterrows():
            current_f1 = float(
                row[
                    "val_macro_f1"
                ]
            )

            if current_f1 > best_f1:
                best_f1 = current_f1
                wait = 0

            else:
                wait += 1

            if wait >= PATIENCE:
                return True

        last_epoch = int(
            history_df.iloc[-1][
                "epoch"
            ]
        )

        if last_epoch >= EPOCHS:
            return True

        return False

    except Exception:
        return False

def train_one_model(
    short_name,
    timm_model_name
):
    seed_all_compare()

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    batch_size = BATCH_BY_MODEL[
        short_name
    ]

    accumulation_steps = max(
        1,
        math.ceil(
            EFFECTIVE_BATCH
            /
            batch_size
        )
    )

    train_loader, val_loader = (
        make_train_val_loaders(
            batch_size
        )
    )

    model = Stage3EfficientNetB0BxSE(
        num_classes,
        LOCAL_STAGE1_W,
        timm_model_name,
        mask_pretrained=True
    ).to(
        device
    )

    if torch.cuda.is_available():
        model = model.to(
            memory_format=torch.channels_last
        )

    aux_head = nn.Linear(
        model.MASK_DIM,
        num_classes
    ).to(
        device
    )

    optimizer = optim.AdamW(
        [
            {
                "params":
                    model.mask_branch.parameters(),
                "lr":
                    LR
            },
            {
                "params":
                    model.se.parameters(),
                "lr":
                    LR
            },
            {
                "params":
                    model.head.parameters(),
                "lr":
                    LR
            },
            {
                "params":
                    aux_head.parameters(),
                "lr":
                    LR
            }
        ],
        weight_decay=WEIGHT_DECAY
    )

    scheduler = (
        optim.lr_scheduler.CosineAnnealingLR(
            optimizer,
            T_max=EPOCHS,
            eta_min=1e-7
        )
    )

    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=USE_AMP
    )

    drive_checkpoint = os.path.join(
        DRIVE_COMPARE_DIR,
        f"stage3_{short_name}_best.pth"
    )

    drive_history = os.path.join(
        DRIVE_COMPARE_DIR,
        f"stage3_{short_name}_history.csv"
    )

    complete_marker = os.path.join(
        DRIVE_COMPARE_DIR,
        f"stage3_{short_name}_COMPLETE.txt"
    )

    best_val_macro_f1 = -1.0
    best_epoch = -1
    wait = 0
    history = []

    print(
        "\n"
        + "=" * 90
    )

    print(
        f"MODEL: {short_name}\n"
        f"EfficientNet: {timm_model_name}\n"
        f"Epochs: {EPOCHS}\n"
        f"Patience: {PATIENCE}"
    )

    print(
        "=" * 90
    )

    for epoch in range(
        1,
        EPOCHS + 1
    ):
        model.train()
        model.rgb_branch.eval()
        aux_head.train()

        optimizer.zero_grad(
            set_to_none=True
        )

        train_losses = []
        train_main_losses = []
        train_labels = []
        train_predictions = []

        train_pbar = tqdm(
            train_loader,
            total=len(
                train_loader
            ),
            desc=(
                f"{short_name} "
                f"Epoch {epoch:02d}/{EPOCHS}"
            ),
            unit="batch",
            dynamic_ncols=True,
            leave=True
        )

        for step, (
            rgb,
            mask,
            labels
        ) in enumerate(
            train_pbar,
            start=1
        ):
            rgb = rgb.to(
                device,
                non_blocking=True
            )

            mask = mask.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            if torch.cuda.is_available():
                rgb = rgb.contiguous(
                    memory_format=torch.channels_last
                )

                mask = mask.contiguous(
                    memory_format=torch.channels_last
                )

            if MASK_DROP_P > 0:
                keep = (
                    torch.rand(
                        mask.size(0),
                        1,
                        1,
                        1,
                        device=device
                    )
                    >
                    MASK_DROP_P
                ).float()

                mask_train = (
                    mask
                    *
                    keep
                )

            else:
                mask_train = mask

            with torch.amp.autocast(
                device_type=AMP_DEVICE,
                enabled=USE_AMP
            ):
                (
                    logits,
                    mask_features
                ) = model.forward_with_parts(
                    rgb,
                    mask_train
                )

                mask_logits = aux_head(
                    mask_features
                )

                main_loss = criterion(
                    logits,
                    labels
                )

                vessel_loss = aux_criterion(
                    mask_logits,
                    labels
                )

                total_loss = (
                    main_loss
                    +
                    AUX_WEIGHT
                    *
                    vessel_loss
                )

            scaler.scale(
                total_loss
                /
                accumulation_steps
            ).backward()

            should_update = (
                step
                %
                accumulation_steps
                ==
                0
                or
                step
                ==
                len(train_loader)
            )

            if should_update:
                scaler.unscale_(
                    optimizer
                )

                torch.nn.utils.clip_grad_norm_(
                    list(
                        model.mask_branch.parameters()
                    )
                    +
                    list(
                        model.se.parameters()
                    )
                    +
                    list(
                        model.head.parameters()
                    )
                    +
                    list(
                        aux_head.parameters()
                    ),
                    max_norm=5.0
                )

                scaler.step(
                    optimizer
                )

                scaler.update()

                optimizer.zero_grad(
                    set_to_none=True
                )

            train_losses.append(
                float(
                    total_loss.detach()
                    .cpu()
                    .item()
                )
            )

            train_main_losses.append(
                float(
                    main_loss.detach()
                    .cpu()
                    .item()
                )
            )

            train_labels.extend(
                labels.detach()
                .cpu()
                .numpy()
            )

            train_predictions.extend(
                logits.argmax(
                    dim=1
                )
                .detach()
                .cpu()
                .numpy()
            )

            train_pbar.set_postfix(
                total_loss=
                    f"{np.mean(train_losses):.4f}",
                main_loss=
                    f"{np.mean(train_main_losses):.4f}",
                patience=
                    f"{wait}/{PATIENCE}"
            )

        train_pbar.close()

        validation = evaluate_with_loss(
            model,
            val_loader
        )

        train_accuracy = accuracy_score(
            train_labels,
            train_predictions
        )

        train_macro_f1 = f1_score(
            train_labels,
            train_predictions,
            average="macro",
            zero_division=0
        )

        train_loss = float(
            np.mean(
                train_losses
            )
        )

        train_main_loss = float(
            np.mean(
                train_main_losses
            )
        )

        history.append(
            {
                "epoch":
                    epoch,

                "train_loss":
                    train_loss,

                "train_main_loss":
                    train_main_loss,

                "val_loss":
                    validation[
                        "loss"
                    ],

                "train_accuracy":
                    train_accuracy,

                "train_macro_f1":
                    train_macro_f1,

                "val_accuracy":
                    validation[
                        "accuracy"
                    ],

                "val_macro_f1":
                    validation[
                        "macro_f1"
                    ],

                "val_macro_auc":
                    validation[
                        "macro_auc"
                    ],

                "val_min_recall":
                    float(
                        np.min(
                            validation[
                                "recall"
                            ]
                        )
                    )
            }
        )

        pd.DataFrame(
            history
        ).to_csv(
            drive_history,
            index=False
        )

        print(
            f"{short_name} | "
            f"Epoch {epoch:02d}/{EPOCHS} | "
            f"Train Total Loss={train_loss:.4f} | "
            f"Train Main Loss={train_main_loss:.4f} | "
            f"Val Loss={validation['loss']:.4f} | "
            f"Val Accuracy={validation['accuracy']:.4f} | "
            f"Val Macro-F1={validation['macro_f1']:.4f} | "
            f"Val AUROC={validation['macro_auc']:.4f}"
        )

        if (
            validation[
                "macro_f1"
            ]
            >
            best_val_macro_f1
        ):
            best_val_macro_f1 = (
                validation[
                    "macro_f1"
                ]
            )

            best_epoch = epoch
            wait = 0

            torch.save(
                model.state_dict(),
                drive_checkpoint
            )

            print(
                "Best validation Macro-F1 checkpoint saved."
            )

        else:
            wait += 1

            print(
                f"Early stopping patience: "
                f"{wait}/{PATIENCE}"
            )

            if wait >= PATIENCE:
                print(
                    f"{short_name}: Early stopping."
                )

                break

        scheduler.step()

    model.load_state_dict(
        torch.load(
            drive_checkpoint,
            map_location=device
        )
    )

    model.eval()

    validation = evaluate_with_loss(
        model,
        val_loader
    )

    print(
        "\n"
        + "-" * 90
    )

    print(
        f"{short_name} VALIDATION RESULTS"
    )

    print(
        "-" * 90
    )

    print(
        f"Validation Loss        = "
        f"{validation['loss']:.4f}\n"
        f"Validation Accuracy    = "
        f"{validation['accuracy']:.4f}\n"
        f"Validation Macro-F1    = "
        f"{validation['macro_f1']:.4f}\n"
        f"Validation Weighted-F1 = "
        f"{validation['weighted_f1']:.4f}\n"
        f"Validation Macro-AUROC = "
        f"{validation['macro_auc']:.4f}"
    )

    result = {
        "model":
            short_name,

        "timm_model":
            timm_model_name,

        "mask_dim":
            model.MASK_DIM,

        "fuse_dim":
            model.FUSE_DIM,

        "batch_size":
            batch_size,

        "accum_steps":
            accumulation_steps,

        "best_epoch":
            best_epoch,

        "val_loss":
            validation[
                "loss"
            ],

        "val_accuracy":
            validation[
                "accuracy"
            ],

        "val_macro_f1":
            validation[
                "macro_f1"
            ],

        "val_weighted_f1":
            validation[
                "weighted_f1"
            ],

        "val_macro_auc":
            validation[
                "macro_auc"
            ],

        "checkpoint":
            drive_checkpoint
    }

    for (
        class_index,
        class_name
    ) in enumerate(
        CLASS_NAMES
    ):
        result[
            f"val_recall_{class_name}"
        ] = validation[
            "recall"
        ][
            class_index
        ]

        result[
            f"val_precision_{class_name}"
        ] = validation[
            "precision"
        ][
            class_index
        ]

    with open(
        complete_marker,
        "w"
    ) as f:
        f.write(
            f"{short_name} completed\n"
        )

        f.write(
            f"patience={PATIENCE}\n"
        )

        f.write(
            f"best_epoch={best_epoch}\n"
        )

        f.write(
            f"best_val_macro_f1="
            f"{best_val_macro_f1:.8f}\n"
        )

    del model
    del aux_head
    del optimizer
    del scheduler
    del scaler
    del train_loader
    del val_loader

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result

def evaluate_existing_checkpoint(
    short_name,
    timm_model_name,
    checkpoint_path
):
    print(
        f"\nLoading saved {short_name} checkpoint..."
    )

    batch_size = BATCH_BY_MODEL[
        short_name
    ]

    _, val_loader = (
        make_train_val_loaders(
            batch_size
        )
    )

    model = Stage3EfficientNetB0BxSE(
        num_classes,
        LOCAL_STAGE1_W,
        timm_model_name,
        mask_pretrained=False
    ).to(
        device
    )

    if torch.cuda.is_available():
        model = model.to(
            memory_format=torch.channels_last
        )

    model.load_state_dict(
        torch.load(
            checkpoint_path,
            map_location=device
        )
    )

    model.eval()

    validation = evaluate_with_loss(
        model,
        val_loader
    )

    history_path = os.path.join(
        DRIVE_COMPARE_DIR,
        f"stage3_{short_name}_history.csv"
    )

    best_epoch = -1

    if os.path.exists(
        history_path
    ):
        history_df = pd.read_csv(
            history_path
        )

        if len(history_df) > 0:
            best_index = history_df[
                "val_macro_f1"
            ].idxmax()

            best_epoch = int(
                history_df.loc[
                    best_index,
                    "epoch"
                ]
            )

    result = {
        "model":
            short_name,

        "timm_model":
            timm_model_name,

        "mask_dim":
            model.MASK_DIM,

        "fuse_dim":
            model.FUSE_DIM,

        "batch_size":
            batch_size,

        "accum_steps":
            max(
                1,
                math.ceil(
                    EFFECTIVE_BATCH
                    /
                    batch_size
                )
            ),

        "best_epoch":
            best_epoch,

        "val_loss":
            validation[
                "loss"
            ],

        "val_accuracy":
            validation[
                "accuracy"
            ],

        "val_macro_f1":
            validation[
                "macro_f1"
            ],

        "val_weighted_f1":
            validation[
                "weighted_f1"
            ],

        "val_macro_auc":
            validation[
                "macro_auc"
            ],

        "checkpoint":
            checkpoint_path
    }

    for (
        class_index,
        class_name
    ) in enumerate(
        CLASS_NAMES
    ):
        result[
            f"val_recall_{class_name}"
        ] = validation[
            "recall"
        ][
            class_index
        ]

        result[
            f"val_precision_{class_name}"
        ] = validation[
            "precision"
        ][
            class_index
        ]

    print(
        f"\n{short_name} SAVED CHECKPOINT RESULTS"
    )

    print(
        f"Validation Loss        = "
        f"{validation['loss']:.4f}"
    )

    print(
        f"Validation Accuracy    = "
        f"{validation['accuracy']:.4f}"
    )

    print(
        f"Validation Macro-F1    = "
        f"{validation['macro_f1']:.4f}"
    )

    print(
        f"Validation Weighted-F1 = "
        f"{validation['weighted_f1']:.4f}"
    )

    print(
        f"Validation Macro-AUROC = "
        f"{validation['macro_auc']:.4f}"
    )

    del model
    del val_loader

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    return result

results = []

progress_csv = os.path.join(
    DRIVE_COMPARE_DIR,
    "B3_progress.csv"
)

for (
    short_name,
    timm_model_name
) in MODELS.items():

    print(
        "\n"
        + "=" * 100
    )

    print(
        f"CHECKING {short_name}"
    )

    print(
        "=" * 100
    )

    checkpoint_path = os.path.join(
        DRIVE_COMPARE_DIR,
        f"stage3_{short_name}_best.pth"
    )

    completed = history_completed(
        short_name
    )

    if (
        os.path.exists(
            checkpoint_path
        )
        and
        completed
    ):
        print(
            f"{short_name} already trained."
        )

        print(
            "Saved checkpoint found:"
        )

        print(
            checkpoint_path
        )

        print(
            f"Completed PATIENCE={PATIENCE} history found."
        )

        print(
            f"{short_name} training skipped."
        )

        print_saved_training_history(
            short_name
        )

        result = evaluate_existing_checkpoint(
            short_name,
            timm_model_name,
            checkpoint_path
        )

        results.append(
            result
        )

    else:
        if os.path.exists(
            checkpoint_path
        ):
            print(
                f"{short_name} checkpoint exists, "
                "but validation-loss history is incomplete."
            )

            print(
                f"Restarting {short_name} training."
            )

        else:
            print(
                f"{short_name} trained weights not found."
            )

            print(
                f"Starting {short_name} training."
            )

        result = train_one_model(
            short_name,
            timm_model_name
        )

        results.append(
            result
        )

    pd.DataFrame(
        results
    ).to_csv(
        progress_csv,
        index=False
    )

    print(
        "\nProgress saved:"
    )

    print(
        progress_csv
    )

comparison_df = pd.DataFrame(
    results
)

comparison_df = (
    comparison_df
    .drop_duplicates(
        subset=[
            "model"
        ],
        keep="last"
    )
    .reset_index(
        drop=True
    )
)

expected_models = {"B3"}

found_models = set(
    comparison_df[
        "model"
    ].astype(str)
)

missing_models = (
    expected_models
    -
    found_models
)

if missing_models:
    raise RuntimeError(
        f"Missing models: "
        f"{sorted(missing_models)}"
    )

comparison_csv = os.path.join(
    DRIVE_COMPARE_DIR,
    "B3_validation_summary.csv"
)

comparison_df.to_csv(
    comparison_csv,
    index=False
)

ranked_df = (
    comparison_df
    .sort_values(
        by=[
            "val_macro_f1",
            "val_accuracy",
            "val_macro_auc"
        ],
        ascending=[
            False,
            False,
            False
        ]
    )
    .reset_index(
        drop=True
    )
)

print(
    "\n"
    + "=" * 110
)

print(
    "FIXED B3 GAN-REFINED VALIDATION SUMMARY"
)

print(
    "=" * 110
)

print(
    ranked_df[
        [
            "model",
            "mask_dim",
            "best_epoch",
            "val_loss",
            "val_accuracy",
            "val_macro_f1",
            "val_weighted_f1",
            "val_macro_auc"
        ]
    ].to_string(
        index=False,
        float_format=lambda x:
            f"{x:.4f}"
    )
)

best = ranked_df.iloc[0]

BEST_MODEL = str(
    best[
        "model"
    ]
)

BEST_TIMM_MODEL = str(
    best[
        "timm_model"
    ]
)

BEST_CHECKPOINT = str(
    best[
        "checkpoint"
    ]
)

selected_validation_result = {
    "selected_model":
        BEST_MODEL,

    "timm_model":
        BEST_TIMM_MODEL,

    "batch_size":
        int(
            best[
                "batch_size"
            ]
        ),

    "best_epoch":
        int(
            best[
                "best_epoch"
            ]
        ),

    "validation_loss":
        float(
            best[
                "val_loss"
            ]
        ),

    "validation_accuracy":
        float(
            best[
                "val_accuracy"
            ]
        ),

    "validation_macro_f1":
        float(
            best[
                "val_macro_f1"
            ]
        ),

    "validation_weighted_f1":
        float(
            best[
                "val_weighted_f1"
            ]
        ),

    "validation_macro_auc":
        float(
            best[
                "val_macro_auc"
            ]
        ),

    "checkpoint":
        BEST_CHECKPOINT
}

selected_validation_csv = os.path.join(
    DRIVE_COMPARE_DIR,
    "SELECTED_MODEL_VALIDATION.csv"
)

pd.DataFrame(
    [
        selected_validation_result
    ]
).to_csv(
    selected_validation_csv,
    index=False
)

print(
    "\n"
    + "=" * 110
)

print(
    "SELECTED MODEL BASED ON VALIDATION"
)

print(
    "=" * 110
)

print(
    f"Model: {BEST_MODEL}"
)

print(
    f"Best Epoch: "
    f"{int(best['best_epoch'])}"
)

print(
    f"Validation Loss: "
    f"{best['val_loss']:.4f}"
)

print(
    f"Validation Macro-F1: "
    f"{best['val_macro_f1']:.4f}"
)

print(
    f"Validation Accuracy: "
    f"{best['val_accuracy']:.4f}"
)

print(
    f"Validation AUROC: "
    f"{best['val_macro_auc']:.4f}"
)

print(
    "Checkpoint:"
)

print(
    BEST_CHECKPOINT
)

print(
    "\n"
    + "=" * 110
)

print(
    "B3 TRAINING AND VALIDATION COMPLETED"
)

print(
    "=" * 110
)

print(
    "\nNO TEST DATA WAS USED."
)

print(
    "\nComparison CSV:"
)

print(
    comparison_csv
)



print(
    selected_validation_csv
)
```
